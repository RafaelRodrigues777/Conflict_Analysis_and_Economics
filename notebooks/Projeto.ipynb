```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Carregar os dados
try:
    df_worldbank = pd.read_excel("../data/WorldBank.xlsx")
except FileNotFoundError:
    print("Erro: O arquivo '../data/WorldBank.xlsx' não foi encontrado.")
    df_worldbank = pd.DataFrame()

if not df_worldbank.empty:
    # Agregar por País e Ano para garantir dados consolidados
    colunas_agg = {
        "Death rate, crude (per 1,000 people)": "mean",
        "GDP (USD)": "mean",
    }
    df_consolidado = (
        df_worldbank.groupby(["Country Name", "Year"])
        .agg(colunas_agg)
        .dropna()
        .reset_index()
    )

    # 2. IDENTIFICAÇÃO AUTOMÁTICA DOS PAÍSES
    # Descobre o ano mais antigo disponível nos dados filtrados
    ano_mais_antigo = df_consolidado["Year"].min()

    # Filtra os dados apenas desse ano antigo
    df_ano_antigo = df_consolidado[df_consolidado["Year"] == ano_mais_antigo]

    # Seleciona os 5 países com maior taxa de morte nesse ano de partida
    top_paises_mortalidade = df_ano_antigo.nlargest(
        5, "Death rate, crude (per 1,000 people)"
    )["Country Name"].tolist()

    print(f"Ano base identificado: {int(ano_mais_antigo)}")
    print(f"Países com maior mortalidade selecionados: {top_paises_mortalidade}")

    # Filtrar o DataFrame consolidado para conter apenas o histórico desses 5 países
    df_filtrado = df_consolidado[
        df_consolidado["Country Name"].isin(top_paises_mortalidade)
    ]

    # 3. Construir o gráfico de trajetórias
    plt.figure(figsize=(14, 8))

    # Mapeamento dinâmico de cores usando um mapa de cores (Colormap) do Matplotlib
    cmap = plt.get_cmap("tab10")

    for i, pais in enumerate(top_paises_mortalidade):
        df_pais = df_filtrado[df_filtrado["Country Name"] == pais].sort_values(
            "Year"
        )

        if not df_pais.empty:
            x = df_pais["Death rate, crude (per 1,000 people)"]
            y = df_pais["GDP (USD)"] / 1_000_000_000  # Convertendo para Bilhões de USD
            cor = cmap(i)

            # Plota a linha conectando a evolução cronológica
            plt.plot(x, y, linestyle="-", alpha=0.4, color=cor)

            # Plota os pontos de cada ano
            plt.scatter(x, y, s=40, alpha=0.6, color=cor)

            # Destaca a bolha do ANO MAIS RECENTE histórica do país
            plt.scatter(
                x.iloc[-1],
                y.iloc[-1],
                s=180,
                color=cor,
                label=pais,
                edgecolors="black",
                zorder=5,
            )

            # Texto do Ano Inicial (ponto de partida)
            plt.text(
                x.iloc[0],
                y.iloc[0],
                f" {int(df_pais['Year'].iloc[0])}",
                fontsize=9,
                alpha=0.8,
            )
            # Texto do Ano Final (ponto mais recente)
            plt.text(
                x.iloc[-1],
                y.iloc[-1],
                f"  {int(df_pais['Year'].iloc[-1])}",
                fontsize=11,
                weight="bold",
            )

    # 4. Customizações e Eixo Invertido para leitura correta do progresso humano
    plt.title(
        f"Evolução dos Países com Maior Taxa de Mortalidade em {int(ano_mais_antigo)}\n(Crescimento do PIB vs. Queda da Mortalidade)",
        fontsize=14,
        weight="bold",
        pad=15,
    )
    plt.xlabel("Taxa de Mortalidade Bruta (por 1.000 pessoas) → Menor é melhor")
    plt.ylabel("PIB (Em Bilhões de USD)")

    # IMPORTANTE: Inverte o eixo X. O país progride andando para a DIREITA e para o TOPO.
    plt.gca().invert_xaxis()

    # Formata o eixo Y para evitar o formato de notação científica confusa
    plt.ticklabel_format(style="plain", axis="y")

    plt.grid(True, linestyle="--", alpha=0.4)
    plt.legend(title="Países Analisados", loc="upper left", fontsize=11)
    plt.tight_layout()
    plt.show()
```
