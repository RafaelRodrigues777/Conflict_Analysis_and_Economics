```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.pyplot as plt
import pandas as pd

# 1. Criar um dataset fictício de exemplo com Pandas
# Substitua esta parte pelo carregamento do seu arquivo (ex: pd.read_csv('seu_arquivo.csv'))
data = {
    "Cidade": ["São Paulo", "Rio de Janeiro", "Brasília", "Manaus", "Porto Alegre"],
    "Latitude": [-23.5505, -22.9068, -15.7938, -3.1190, -30.0346],
    "Longitude": [-46.6333, -43.1729, -47.8828, -60.0217, -51.2177],
}
df = df_conflicts

# 2. Configurar a figura do Matplotlib e a projeção do mapa
# Escolhemos a projeção PlateCarree para a exibição, mas poderia ser outra (como Mercator)
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())

# 3. Adicionar elementos de fundo ao mapa (linhas costeiras, fronteiras, etc.)
ax.add_feature(cfeature.LAND, facecolor="lightgray")
ax.add_feature(cfeature.OCEAN, facecolor="lightblue")
ax.add_feature(cfeature.COASTLINE, edgecolor="black", linewidth=0.8)
ax.add_feature(cfeature.BORDERS, linestyle=":", edgecolor="gray")

# 4. Plotar os pontos de Latitude e Longitude do Dataset
# O truque essencial é o "transform=ccrs.PlateCarree()" para coordenadas Lat/Lon
ax.scatter(
    df["longitude"],
    df["latitude"],
    color="red",
    marker="o",
    s=1,  # Tamanho dos pontos
    transform=ccrs.PlateCarree(),
    zorder=5,  # Garante que os pontos fiquem por cima do mapa
)

# 5. Adicionar rótulos de grade (opcional, para mostrar linhas de lat/lon)
ax.gridlines(draw_labels=True, dms=True, x_inline=False, y_inline=False)

# 6. Definir o enquadramento/limites do mapa (opcional - focando na América do Sul)
# ax.set_extent([-80, -30, -40, 10], crs=ccrs.PlateCarree())

# Exibir o gráfico
plt.title("Pontos do Dataset Plotados com Cartopy", fontsize=14)
plt.show()
```
